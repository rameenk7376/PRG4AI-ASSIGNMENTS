```python
sentence=input("Enter a sentence:")
```

```python
def get_words(sentence):
    words=[]
    word=""
    for ch in sentence+" ":
        if ch!=" ":
            word=word+ch
        elif word!="":
            words=words+[word]
            word=""
    return words
```

```python
def count_vowels(sentence):
    count=0
    for ch in sentence:
        if ch in "aeiouAEIOU":
            count=count+1
    return count
```

```python
def f_longest(words):
    longest=words[0]
    for l in words:
        if len(l)>len(longest):
            longest=l
    return longest
```

```python
def f_shortest(words):
    shortest=words[0]
    for s in words:
        if len(s)<len(shortest):
            shortest=s
    return shortest
```
