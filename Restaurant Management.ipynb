```python
menu={
    "1" : "Burger",
    "2" : "Pizza",
    "3" : "Fries",
    "4" : "Drink"
}
```

```python
prices={
    "Burger" : 500,
    "Pizza" : 800,
    "Fries" : 250,
    "Drink" : 150
}
```

```python
def calc_price(item,quantity):
    return prices[item]*quantity
```

```python
print("----------MENU----------")
print("1.Burger Rs.500")
print("1.Pizza Rs.800")
print("1.Fries Rs.250")
print("1.Drink Rs.150")
```

```python
bill=0
order=""
```

```python
while True:
    choice=input("Select item number:")
    if choice in menu:
        item=menu[choice]
        quantity=int(input("Enter quantity:"))
        price=calc_price(item,quantity)
        bill=bill+price
        order=order+item+" x "+str(quantity)+" = Rs."+str(price)+"\n"
    else:
        print("Wrong choice")
    more=input("Want to order more? (yes/no)")
    if more!="yes":
        break
```

```python
print("------YOUR BILL-------")
print(order)
print("Total Bill= Rs.",bill)
```

```python
file=open("orders.txt","w")
file.write("-------ORDER-------")
file.write(order)
file.write("Total Bill= Rs."+str(bill)+"\n")
file.close()
```
