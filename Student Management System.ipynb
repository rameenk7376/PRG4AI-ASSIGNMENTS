```python
students=[]

def calculate(m1,m2,m3):
    total=m1+m2+m3
    percentage=total/300*100

    if percentage>=80 and percentage<=100:
        grade="A"
    elif percentage>=70 and percentage<80:
        grade="B"
    elif percentage>=60 and percentage<70:
        grade="C"
    elif percentage>=50 and percentage<60:
        grade="D"
    else:
        grade="F"

    return total,percentage,grade
```

```python
for i in range(5):
    print("Student",i+1)
    name=input("Enter Student's Name:")
    rollno=input("Enter Student's Roll No.:")
    python=int(input("Enter Python Marks:"))
    ai=int(input("Enter AI Marks:"))
    maths=int(input("Enter Maths Marks:"))
    total, percentage, grade = calculate(python, ai, maths)
    
    student = {
            "name": name,
            "rollno": rollno,
            "python": python,
            "ai": ai,
            "maths": maths,
            "total": total,
            "percentage": percentage,
            "grade": grade
        }
    
    students.append(student) 
```

```python
print("\n--------------STUDENT RESULTS-----------------\n")
for s in students:
    print("\nName:",s["name"])
    print("\nRoll no:",s["rollno"])
    print("\nTotal:",s["total"],"/300")
    print("\nName:",s["percentage"],"%")
    print("Grade:",s["grade"])
```

```python
top=students[0]
for s in students:
    if s["percentage"]>top["percentage"]:
        top=s

print("Top Student: ",top["name"],"\nMarks:",top["percentage"],"%")
```

```python
file=open("results.txt","w")

file.write('-------Student Results------')
for s in students:
    file.write("\n Name:"+s["name"])
    file.write("\n Roll no:"+s["rollno"])
    file.write("\n Total:"+s["total"])
    file.write("\n Percentage:"+s["percentage"])
    file.write("\n Grade:"+s["grade"])

file.close()
```
